# The Best Predictor and The Conditional Expectation Function

## CEF and Linear Regression Estimator

### Best Linear Prediction in Population
- Set Up:
    - Consider a scalar random variable $Y$ as the outcome of interest
    - A p-vector of covariates $X=(X_1, X_2, ..., X_p)$
    - We assume that a constant of 1 is included as the first component in $X$, that is, $X_1 = 1$
- Linear Prediction Rule:
    - The predicted value of Y given X will be of the linear form
    - $\sum_{j=1}^{p} \beta_j X_j \;=\; \beta^\top X, \quad \text{where } \beta = (\beta_1, \ldots, \beta_p)^\top$
    - $\beta$ is called the regression parameter or coefficient
- Best Linear Predictor (BLP):
    - $\beta$ that solves the minimization problem $\min_{b \in \mathbb{R}^p} \; E\!\left[ \left( Y - b^\top X \right)^2 \right]$
    - That is, BLP is the linear prediction rule that minimizes the Mean Squared Error (MSE) at population level
- Property of the BLP:
    - FOC of the minimization problem implies that $E[( Y - \beta^\top X )X]=0$ (also called Normal Equation)
    - By Law of Iterated Expectation (LIE), $E[E[( Y - \beta^\top X )X]|X]=0$
    - Implying that: $E[(E[Y|X] - \beta^\top X )X]=0$
    - This is the FOC of the MSE minimization problem that the BLP of $E[Y|X]$ solves
    - Hence, BLP of $Y$ is also BLP of $E[Y|X]$ (conditional expectation of Y given X)

### Best Prediction in Population
- Set Up:
    - Suppose $W$ are "raw" regressors / features
    - $X$ are constructed regressors of the form $X= T(W) = [T_1(W), T_2(W), ... T_p(W)]^\top$
    - $T(W)$ denotes the set of transformations, sometimes called the dictionary of transformations
    - Example transformations include polynomials, interactions between variables, and applying functions such as the logarithm or exponential
    - This may also include the identity transformation, $T(W) = W$
- Best Predictor:
    - Suppose the best predictor (BP) of Y given W is $m(W)$
    - In the population, BP solves $\min_{m(W)} E[(Y - m(W))^2]$
    - The solution is $g(W) = E[Y|W]$
    - Hence, $E[Y|W]$ (CEF, also called regression function) is the best prediction rule
- Porperty of the BP:
    - It solves the same problem as the best linear prediction rule among a larger class of candidate rules
    - Hence, the conditional expectation function generally provides better predictions than the best linear prediction rule
    - Unless the CEF turns out to be linear, in which case the CEF and best linear prediction rule (in population) coincide
- Connection with BLP:
    - Remember that BLP of Y in X is $\beta$ that solves $\min_{b \in \mathbb{R}^p} \; E\!\left[ \left( Y - b^\top T(W) \right)^2 \right]$
    - It turns out $E\!\left[ \left( Y - b^\top T(W) \right)^2 \right] = E\!\left[ \left( g(W) - b^\top T(W) \right)^2 \right] + E\!\left[ \left( Y - g(W) \right)^2 \right]$
    - $E\!\left[ \left( Y - g(W) \right)^2 \right]$ is a constant that does not depend on $b$
    - Therefore, $b$ that solves $E\!\left[ \left( Y - b^\top T(W) \right)^2 \right]$ also solves $E\!\left[ \left( g(W) - b^\top T(W) \right)^2 \right]$
    - The BLP of Y, $\beta^\top X = \beta^\top T(W)$, is the Best Linear Approximation of CEF $E[Y|W]$ (best predictor)
    - Intuition:
        - The CEF $E[Y|W]$ captures all the predictive information about Y that is contained in W
        - So whether we try to approximate the CEF directly or try to predict Y, we end up with the same linear approximation $\beta^\top T(W)$
    - As the dictionary of transformations $T(W)$ becomes richer, the quality of the approximation of the BLA $\beta^\top T(W)$ to the best predictor $E[Y|W]$ improves
    - With abudant richness of transformations $T(W)$, we can approximate BLP $\beta^\top X = \beta^\top T(W)$ as close as possible to best predictor $E[Y|W]$ in **Population = Sample with Infinite Observations !!!**

### Best Linear Prediction in Finite Samples
- Sample:
    - In practice, we do not access to the entire population, but observes only a sample
    - However, we can construct the in-sample BLP for Y using X analogously to the population case 
    - We achieve this by replacing theoretical expected values, $E$, with empirical averages, $\mathbb{E}_n$
- In-Sample Best Linear Predictor (OLS):
    - The in-sample BLP coefficient $\hat \beta$ solves $\min_{b \in \mathbb{R}^p} \; \mathbb{E}_n\!\left[ \left( Y - b^\top X \right)^2 \right]$
    - This method of derving $\hat \beta$ is also known as Ordinary Least Squares (OLS)
    - That is, $\hat \beta$ minimizes the sample MSE for predicting Y using the linear rule $b^\top X = b^\top T(W)$
    - FOC: $\mathbb{E}_n[( Y - \hat \beta^\top X )X]=0$
- In-sample Properties of OLS:
    - As OLS minimize the in-smaple MSE, it is optimal among linear unbiased estimators for in-sample prediction
    - However, this optimality is not guaranteed to extend to out-of-sample prediction
- Out-of-sample Properties of OLS:
    - The best linear prediction rule in the population is $\beta ^ \top X$
    - As the dictionary of transformation $X = T(W)$ becomes richer, $p$ increases, and $\beta ^ \top X$ will approximate the CEF (best predictor) well
    - The problem is, whether $\hat \beta ^ \top X$ estimates $\beta ^ \top X$ well, especially when $p$ increases and the population BLP gets closers to CEF
    - Short answer: the expected deviation of  $\hat \beta ^ \top X$ from $\beta ^ \top X$ is bounded by $\frac{p}{n}$, where $n$ is number of observations in sample
    - In other word, if n is large and p is small relative to n, in-sample BLP approximates the population BLP well
    - But the good approximation of CEF by population BLP can require large p relative to n
    - In this setting, p/n is not small, in-sample BLP can be substainally different from population BLP and the best predictor (CEF)
    - This setting is usually referred as overfitting, and the limitation of OLS in high-dimensional setting (high p) motivates alternative estimators

### Two Direction of Deviation from OLS
- We have discussed that as p is large relative to n, the in-smaple BLP (OLS estimator) can deviate substantially from population BLP and the CEF
- Two causes of large p (high-dimension setting):
    - The number of covariates (dimension of W) is large
    - The number of transformations (T(W)) required is large for population BLP to approximate CEF well (e.g. CEF is highly non-linear and involves mutiple interactions)
- To overcome the limitation of OLS in this setting, two directions exist:
    - Biased Linear Regression Estimators:
        - Biased estimators like ridge regression can achieve better out-of-sample MSE by trading off some bias for reduced variance
        - This tradeoff becomes especially important in high-dimensional settings where OLS may overfit
    - Non-parametric Estimators:
        - Estimating CEF $g(W)$ can be done using some form of nonparametric regression or smoothing (without making parametric assumptions about $g(X)$)
        - This prevent the large set of transformation required / estimating in high-dimensional space 

## Nonparametric estimators

### Nonparametric v.s. Parametric Approach
- Parametric approach:
    - Assume a functional form for $g(W)$ and estimate parameters
    - Example: linear regression, polynomial regression
- Nonparametric approach:
    - Use flexible algorithms to approximate the CEF without explicit assumption on the functional form of $g(W)$
    - Example: kernel regression, local polynomial regression, random forests, neural networks

### Basic Concepts for Nonparametric Estimators
- Covariate Space $\mathcal{X}$:
    - Let the covariates be the set of variables $X_1, X_2, ..., X_p$
    - The feature space or covariate space, $\mathcal{X}$, denotes the set of all possible combinations of the values that can be taken by $X_1, ...., X_p$
    - The conditional expectation function (CEF) can be calculated as the set of conditional expectations on $\mathcal{X}$
- Partition on $\mathcal{X}$: 
    - A partition is a set of subsets $R_1, ..., R_J$ of the space $\mathcal{X}$ such that $R_i \cap R_j = \varnothing$ and $R_1 \cup R_2 \cup \cdots \cup R_J = \mathcal{X}$
- Tree:
    - A tree $\Pi$ is a partition of the feature space $\mathcal{X}$ such that:
        - $\Pi = \{ I_1, \ldots, I_M \}$
        - $\bigcup_{j=1}^{M} I_j = \mathcal{X}$
    - With M denotes the number of leafs in the partition and $I_j$ denotes the j th leaf

### Conditional Mean Estimator
- Simplified Set Up：
    - Consider that there are only two covariates
    - $X=(X_1, X_2)$, both are discrete
    - the outcome $Y$ is continous
    - the CEF is $E[Y|X] = g(X)$
- Conditional Mean Estimator:
    - The CEF is the set of conditional expectations over the covariate space $\mathcal{X}$
    - Given that $X$ is discrete,  CEF is the collection of $E(Y|X_1, X_2)$ for every distinct combination of $(X_1, X_2)$
    - Then, a natural way to estimate CEF is computing the empirical average of Y for each distinct paired values $(X1, X2)$
    - Formally, the conditional mean estimator is $$ \hat g(X_i) = \mathbb{E}_n(Y_i|X_1 = x_1, X_2 = x_2) = \frac{\sum_{i=1}^{N}1(X_1 = x_1 \cap X_2 = x_2)Y_i}{\sum_{i=1}^{N}1(X_1 = x_1 \cap X_2 = x_2)} $$
- Non-parametric Propery:
    - Estimating CEF as the set of conditional average for every distinct value of X (our partitions), exploits the fact that elements of X are discrete
    - This estimator is non-parametric, as we are not imposing a functional form on the CEF
    - We are letting the data speak without imposing e.g. linearity
- Limitation:
    - This approach is based upon prior knowledge of the partitions (partitions constructed by discrete levels) when the dimension of X is **small** and **discrete**
    - If the dimension is large: there would be uncertainty in terms of how to construct partitions such that the set of average predictions have good properties
    - If any dimension is continous: again uncertainity in how to construct partitions
- Alternative Conditional Mean Estimators:
    - Binned Estimator (h) - h bin width (discontinuous predictor)
    - Nearest Neighbour (k) - k nearest neighbours (discontinuous predictor)
    - Moving Average Estimator (h) - h interval width (discontinuous predictor)
    - Kernel Estimator (h) - h controls weight (continuous predictor)
    - Tree-based Estimators (discontinuous predictor)
    - These approaches povide sensible ways to construct partions when either dimension of X is **large** or **continous**

### Binned Estimator
- Divide data into a fixed number of discrete bins
- The bin width h is the only hyperparameter: wide loses detail, narrow gets noisy
- Compute statistics within bins (means, medians etc.)
- Histogram is a special case of binned estimator:
    - It counts observations in each bin
    - It faces same trade-off as binned estimator

### Nearest Neighbour Estimator
- Estimator:
    - Hyperparameter k controls the number of “local observations”
    - The prediction of Y at X is the average of k closest observations to X in the sample
    - It is also called k nearest neighbour estimator (KNN)
- Porperties:
    - The distance of observations (used to compute conditional average) to x is variable, but the number of observations used is fixed
    - The flip side of K-NN is the moving average estimator

### Moving Average Estimator
- Estimator:
    - Hyperparameter h controls the distance of observations (used to compute conditional average) to x
    - Consider all observations with $|X_i − X|≤ h$, calculate average of these observations as prediction of Y for given X
    - Formally: 
    $$ \hat{\mu}(x)
    = \frac{\sum_{i=1}^{N} \mathbf{1}\bigl( |x - x_i| \le h \bigr) y_i}
    {\sum_{i=1}^{N} \mathbf{1}\bigl( |x - x_i| \le h \bigr)} 
    = \sum_i w_i y_i $$
    - where 
    $$ w_i = \frac{\mathbf{1}\bigl( |x - x_i| \le h \bigr)}
    {\sum_{i=1}^{N} \mathbf{1}\bigl( |x - x_i| \le h \bigr)} $$
- Properties:
    - Number of observations within each interval is variable, but distance fixed
    - In sparse region, the estimator is subject to noise:
        - In comparision, K-NN adapts to local data density, intervals get wider in sparse regions
        - Adaptiveness comes from the varying width needed to capture that fixed number of points across dense and sparse regions
    - The estimator is a discontinuous function:
        - This is due to the use of the indicator function $1(.)$ to construct weights / prediction
        - Observations that are close but outside of either end point, will receive a weight that is zero
        - K-NN have same problem
        - Kernel estimator circumvents this problem by assigning a weight which is dependent on the closeness to X

### Kernel Estimator
- Nadaraya-Watson Estimator is a Kernel Estimator
- Formally: 
    $$ \hat{g}(x)
    = \frac{\sum_{i=1}^{N} K( \frac{x - x_i} {h} ) y_i}
    {\sum_{i=1}^{N} K( \frac{x - x_i} {h} )} 
    = \sum_i W_i(x_i) y_i $$
- Where:
    $$ W_i(x_i) = \frac{K( \frac{x - x_i} {h} )}
    {\sum_{i=1}^{N} K( \frac{x - x_i} {h} )} $$
- The estimator is the weighted sum of the $Y_i$:
    - With weights determined by the distance of individual $x_i$ from the reference point $x$
    - The weight is decreasing in distance between $x_i$ and reference point $x$
    - $h$ controls how quickly weights decay with distance

### Common Challenges of Non-parametric Estimators
- How to choose the optimal tuning parameter:
    - Nearest neighbour k, bin width h, or interval width h
    - How to balance the trade-off between noise and loss of information
- How to handle the bias-variance tradeoff
- How to deal with the curse of dimensionality in higher dimensions
- Random Forest Estimator tackle some of these challenges

### Random Forests Estimator
- Not a single partition like binning, they average the predictions over many different partitions (trees)
- Each tree (partition) sees a random subset of the data and features, averaging over them reduce variance of the estimator
- Adaptive rather than fixed partitioning: splits are chosen adaptively based on reducing error

### Ensembling Different Estimators
- Definition:
    - Ensemble learning refers to combining multiple models to improve predictive performance
    - This can be done by model averaging
- Set Up:
    - Consider 3 estimators:
        - Random forest: $Y^{RF}$
        - Neural Network: $Y^{NN}$
        - LASSO: $Y^{LASSO}$
    - The predictions can be averaged using weights $p^{RF}$, $p^{RF}$, $p^{RF}$
        - Chosen through minimizing cross-validation error
        - $p^{RF} + p^{RF} + p^{RF} = 1$
        - $p^{RF}>0, p^{RF}>0,  p^{RF}>0$
- Advantage of ensembling:
    - Accounts for model uncertainty (reduce variance through averaging) 
    - Combines complementary strengths of models
- Models have complementary strength:
    - Random Forest: nonparametric, can handle irrelevant features, nonlinearities, and interactions
    - Neural Network: highly flexible functional form, many layers and nodes
    - LASSO: linear model with regularization, perfrom well when coefficient space is sparse
- Ensembles usually outperform any single model:
    - Formal results comparing ensemble to single models are limited
    - But empirical performance is strong motivation

## FWL Theorem

### Orthogonality
- Two vectors $X$ and $Y$ in a vector space are orthogonal if their inner product equals zero: $X ^\top Y = 0$
- In summation notation: $\sum_{i=1}^n X_iY_i = 0$
- In the context of random variables $X$ and $Y$:
    - $X$ and $Y$ are orthogoal if $E[X ^\top Y] = E[\sum_{i=1}^n X_iY_i] = 0$
    - In regression analysis, residuals $ε$ are orthogonal to regressors $X$ if $X^\topε = 0$

### FWL Theorem Introduction
- The Frisch-Waugh-Lovell (FWL) theorem is based on the idea of partialling out
- Partialling out: the process of isolating the variation in one variable that is **orthogonal** (not linearlly predictable by) to another variable
- The FWL theorem allows us to estimate coefficients of a multivariate regression through running several univariate regressions
- Example: 
    - Suppose we have $X_1$, $X_2$, and $Y$
    - The FWL theorem asserts that we can measure the predictive effect of $X_1$ on $Y$
    - After partialling out the linear predictive effect of $X_2$ on $X_1$ and $X_2$ on $Y$

### FWL Theorem in Formal
- Consider the regression: $Y = \alpha + \beta _1 X_1 + \beta _2 X_2 + ε$
- The FWL theorem states that coefficient $\beta _2$ can be obtained from the regression:
    - $M_{X_2}Y = M_{X_2}X_1 \beta _2 + M_{X_2} ε$
    - Where $M_{X_2} = I - X_2(X_2 ^\top X_2)^{-1}X_2 ^\top$
    - $M_{X_2}$ denotes an orthogonal projection matrix
    - It projects a vector onto the orthogonal complement of the column space of $X_2$ (isolate variation that cannot be explained by $X_2$)
- Intuitively:
    - When we apply $M_{X_2}$ to any variable, we are removing the linear predictive effecr of $X_2$ on that variable
    - Mathematically, we’re extracting the residuals after regressing that variable on $X_2$
    - Therefore, regressing $M_{X_2}Y$ on $M_{X_2}X_1$ is regressing the residualized $Y$ on the residualized $X_1$
    - This isolates the partial effect of $X_1$ on $Y$ that is not explained by $X_2$ (as both $M_{X_2}Y$ and $M_{X_2}X_1$ are orthogoal to $X_2$)
- Application:
    - In OLS, FWL seems redudant, as we can estimate multi-linear regression model directly (do not need two univariate regression)
    - However, a version of the FWL (discussed later) is crucial to connect ML with causal inference
